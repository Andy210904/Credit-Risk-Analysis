```python
import pandas as pd

# -----------------------------
# (All the rule functions here — same as before)
# -----------------------------
def classify_ebitda_margin(x):
    if pd.isna(x): return "Medium"
    if x >= 20: return "High"
    elif x < 5: return "Low"
    else: return "Medium"

def classify_ebit_margin(x):
    if pd.isna(x): return "Medium"
    if x >= 15: return "High"
    elif x < 3: return "Low"
    else: return "Medium"

def classify_debt_to_equity(x):
    if pd.isna(x): return "Medium"
    if x <= 1: return "High"
    elif x > 2.5: return "Low"
    else: return "Medium"

def classify_interest_coverage(x):
    if pd.isna(x): return "Medium"
    if x >= 5: return "High"
    elif x < 1.5: return "Low"
    else: return "Medium"

def classify_dscr(x):
    if pd.isna(x): return "Medium"
    if x >= 2: return "High"
    elif x < 1: return "Low"
    else: return "Medium"

def classify_current_ratio(x):
    if pd.isna(x): return "Medium"
    if x >= 1.5: return "High"
    elif x < 1: return "Low"
    else: return "Medium"

def classify_quick_ratio(x):
    if pd.isna(x): return "Medium"
    if x >= 1: return "High"
    elif x < 0.5: return "Low"
    else: return "Medium"

def classify_dso_days(x):
    if pd.isna(x): return "Medium"
    if x <= 30: return "High"
    elif x > 90: return "Low"
    else: return "Medium"

def classify_dpo_days(x):
    if pd.isna(x): return "Medium"
    if x >= 60: return "High"
    elif x < 20: return "Low"
    else: return "Medium"

def classify_dio_days(x):
    if pd.isna(x): return "Medium"
    if x <= 60: return "High"
    elif x > 150: return "Low"
    else: return "Medium"

def classify_revenue_cagr(x):
    if pd.isna(x): return "Medium"
    if x >= 10: return "High"
    elif x < 0: return "Low"
    else: return "Medium"

def classify_years_in_operation(x):
    if pd.isna(x): return "Medium"
    if x >= 10: return "High"
    elif x < 3: return "Low"
    else: return "Medium"

def classify_ownership_type(x):
    if pd.isna(x): return "Medium"
    return "High" if str(x).lower() == "listed" else "Medium"

def classify_auditor_tier(x):
    if pd.isna(x): return "Medium"
    return "High" if x == 1 else "Low"

def classify_governance_score(x):
    if pd.isna(x): return "Medium"
    if x >= 70: return "High"
    elif x < 40: return "Low"
    else: return "Medium"

def classify_esg_controversies(x):
    if pd.isna(x): return "Medium"
    if x == 0: return "High"
    elif x >= 3: return "Low"
    else: return "Medium"

def classify_country_risk(x):
    if pd.isna(x): return "Medium"
    if x <= 30: return "High"
    elif x > 70: return "Low"
    else: return "Medium"

def classify_industry_cyclicality(x):
    if pd.isna(x): return "Medium"
    if x == 0: return "High"
    elif x == 2: return "Low"
    else: return "Medium"

def classify_fx_revenue(x):
    if pd.isna(x): return "Medium"
    if x <= 30: return "High"
    elif x > 70: return "Low"
    else: return "Medium"

def classify_hedging_policy(x):
    if pd.isna(x): return "Medium"
    return "High" if str(x).lower() == "yes" else "Low"

def classify_covenant_quality(x):
    if pd.isna(x): return "Medium"
    return "High" if x == 1 else "Low"

def classify_payment_incidents(x):
    if pd.isna(x): return "Medium"
    return "High" if x == 0 else "Low"

def classify_legal_disputes(x):
    if pd.isna(x): return "Medium"
    return "High" if x == 0 else "Low"

def classify_sanctions_exposure(x):
    if pd.isna(x): return "Medium"
    return "High" if x == 0 else "Low"

def classify_financials_audited(x):
    if pd.isna(x): return "Medium"
    return "High" if str(x).lower() == "yes" else "Low"

def classify_pd(x):
    if pd.isna(x): return "Medium"
    if x <= 1: return "High"
    elif x > 5: return "Low"
    else: return "Medium"

def classify_lgd(x):
    if pd.isna(x): return "Medium"
    if x <= 30: return "High"
    elif x > 60: return "Low"
    else: return "Medium"
# -----------------------------
# Apply all rules to a DataFrame
# -----------------------------

def transform_dataset(df):
    rule_funcs = {
        "ebitda_margin_pct": classify_ebitda_margin,
        "ebit_margin_pct": classify_ebit_margin,
        "debt_to_equity": classify_debt_to_equity,
        "interest_coverage": classify_interest_coverage,
        "dscr": classify_dscr,
        "current_ratio": classify_current_ratio,
        "quick_ratio": classify_quick_ratio,
        "dso_days": classify_dso_days,
        "dpo_days": classify_dpo_days,
        "dio_days": classify_dio_days,
        "revenue_cagr_3y_pct": classify_revenue_cagr,
        "years_in_operation": classify_years_in_operation,
        "ownership_type": classify_ownership_type,
        "auditor_tier": classify_auditor_tier,
        "governance_score_0_100": classify_governance_score,
        "esg_controversies_3y": classify_esg_controversies,
        "country_risk_0_100": classify_country_risk,
        "industry_cyclicality": classify_industry_cyclicality,
        "fx_revenue_pct": classify_fx_revenue,
        "hedging_policy": classify_hedging_policy,
        "covenant_quality": classify_covenant_quality,
        "payment_incidents_12m": classify_payment_incidents,
        "legal_disputes_open": classify_legal_disputes,
        "sanctions_exposure": classify_sanctions_exposure,
        "financials_audited": classify_financials_audited,
        "PD_1y_pct": classify_pd,
        "LGD_pct": classify_lgd,
    }

    for col, func in rule_funcs.items():
        if col in df.columns:
            df[col + "_risk"] = df[col].apply(func)

    # -----------------------------
    # Compute overall risk score
    # -----------------------------
    risk_map = {"High": 2, "Medium": 1, "Low": 0}
    risk_cols = [c for c in df.columns if c.endswith("_risk")]

    df["rule_score"] = df[risk_cols].applymap(lambda x: risk_map.get(x, 1)).sum(axis=1)
    df["risk_gauge"] = (df["rule_score"] / (2 * len(risk_cols))) * 100

    def bucket(score):
        if score < 40: return "High Risk"
        elif score < 70: return "Medium Risk"
        else: return "Low Risk"

    df["risk_bucket_rule_based"] = df["risk_gauge"].apply(bucket)

    return df

# -----------------------------
# Load, transform, and save
# -----------------------------
if __name__ == "__main__":
    # Load your dataset
    df = pd.read_csv("credit_risk_dataset_50_entities.csv")

    # Transform it
    df_transformed = transform_dataset(df)

    # Save into a new CSV
    df_transformed.to_csv("transformed_dataset.csv", index=False)

    print("✅ Transformed dataset saved as 'transformed_dataset.csv'")
```

```python
import pickle

# Load pickled rules
with open("risk_rules_full.pkl", "rb") as f:
    data = pickle.load(f)

rules = data["rules"]
cat_map = data["categorical_mappings"]

def classify_risk(input_data, rules, cat_map):
    classification = {}
    
    for var, val in input_data.items():
        if var not in rules:
            continue
        
        rule = rules[var]
        # Convert categorical input to numeric if needed
        if rule.get("categorical", False):
            val = cat_map[var].get(val, val)  # map word to number
            
        direction = rule["direction"]
        low_cut = rule["low"]
        high_cut = rule["high"]
        
        if direction == "higher":
            if val >= high_cut:
                classification[var+"_risk"] = "High"
            elif val < low_cut:
                classification[var+"_risk"] = "Low"
            else:
                classification[var+"_risk"] = "Medium"
        elif direction == "lower":
            if val <= low_cut:
                classification[var+"_risk"] = "High"
            elif val >= high_cut:
                classification[var+"_risk"] = "Low"
            else:
                classification[var+"_risk"] = "Medium"
    
    # Final evaluation
    total = len(classification)
    high_count = sum(1 for v in classification.values() if v=="High")
    low_count = sum(1 for v in classification.values() if v=="Low")
    if high_count/total >= 0.6:
        classification["final_evaluation"] = "High"
    elif low_count/total >= 0.6:
        classification["final_evaluation"] = "Low"
    else:
        classification["final_evaluation"] = "Medium"
    
    return classification
```
